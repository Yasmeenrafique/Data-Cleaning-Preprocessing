```python
import pandas as pd
import numpy as np
```

Load Dataset

```python
df = pd.read_csv('internship_applications_dirty_dataset.csv')
```

```python
df.head()
```

```python
df.info
```

```python
df.describe()
```

Handle Missing Values

```python
df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
df['Skills'] =df['Skills'].fillna(df['Skills'].mode()[0])
```

Remove Duplicates

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

Detect Outliers

```python
Q1 = df['CGPA'].quantile(0.25)
Q3 = df['CGPA'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['CGPA'] < lower_bound) | (df['CGPA'] > upper_bound)]
```

```python
df = df[(df['CGPA'] >= lower_bound) & (df['CGPA'] <= upper_bound)]
```

Standardize Data

```python
df['Gender'] = df['Gender'].str.capitalize()
```

```python
df
```

```python
df['Name'] = df['Name'].str.strip()
```

```python
df.dtypes
```

```python
df['Age'] =df['Age'].astype(int)
```

```python
df['Experience_Years'] = df['Experience_Years'].apply(lambda x: 'Yes' if x > 0 else 'No')
```

```python
print(df)
```

```python
df.to_csv('internship_applications_cleaned_dataset.csv', index=False)
```
